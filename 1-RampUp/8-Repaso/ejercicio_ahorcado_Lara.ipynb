## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python

import random

def print_rayitas_letras(palabra, letras_adivinadas):
    '''Comprueba cada letra de la palabra contra la lista de letras acertadas'''
    lista = []
    for elemento in palabra:
        if elemento in letras_adivinadas:
            lista.append(elemento)
        else:
            lista.append("_")
    return " ".join(lista)

def comprobar_vidas(contador):
    if contador < 6:
        return True
    else:
        return False


def palabra_aleatoria(lista):
    n_aleatorio =  random.randint(0,len(lista))
    return lista[n_aleatorio]  
    

````

````python
lista_palabras = ["Agua", "café", "té"]
palabra = palabra_aleatoria(lista_palabras).lower() 
# Imprimimos las rayitas iniciales
print("_ " * len(palabra))

errores = 0
letras_adivinadas = [] # Guardamos los aciertos
jugando = True

while jugando:
    if not comprobar_vidas(errores):
        print("Fin del juego, te quedaste sin vidas.")
        jugando = False
        break
            
    letra = input("Introduce una letra: ").lower() 
    
    # Comprobamos siempre contra la palabra original
    if letra in palabra:
        print("Acertaste")
        letras_adivinadas.append(letra) # Guardamos la letra en la memoria
    else:
        print("Fallaste")
        errores = errores + 1
        
    # Imprimimos cómo va el progreso actual
    estado_actual = print_rayitas_letras(palabra, letras_adivinadas)
    print(estado_actual)
    print(f"Errores: {errores}/6\n")
    
    # Comprobar si ya no quedan guiones para ganar el juego
    if "_" not in estado_actual:
        print("¡Felicidades, ganaste!")
        jugando = False
        break
````
