## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time
```

```python

resultados = []
salida = True
while salida == True:
    operacion = input("Introduce una operación: suma, resta, multiplicación, división o potencia, salir: ")
    if operacion == "salir":
            print("Saliendo del programa...devolviendo resultados")
            print(f"Resultados obtenidos: {resultados}")
            salida = False
            break
    else:
        n1 = int(input("Introduce el primer número: "))
        n2 = int(input("Introduce el segundo número: "))
    if operacion == "suma":
        print(f"El resultado de la suma es: {n1+n2}")
        resultados.append(n1+n2)
    elif operacion == "resta":
        print(f"El resultado de la resta es: {n1-n2}")
        resultados.append(n1-n2)
    elif operacion == "multiplicación":
        print(f"El resultado de la multiplicación es: {n1*n2}")
        resultados.append(n1*n2)
    elif operacion == "división":
        if n2 == 0:
            print("No se puede dividir entre 0")
        else:
            print(f"El resultado de la división es: {n1/n2}")
            resultados.append(n1/n2)
    elif operacion == "potencia":
        print(f"El resultado de la potencia es: {n1**n2}")
        resultados.append(n1**n2)
    else:
        print("Operación no válida")
```
