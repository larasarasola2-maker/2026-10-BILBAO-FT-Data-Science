## Ejercicio 11
No queda muy claro cuánto de dispar, de dispersa, es la medida de IBU. Mediante estadística descriptiva (una o dos métricas, sin gráficas), razona la dispersión de la variable IBU.

## Ejercicio 13
Representa el nivel de alcohol (columna `abv`) en un histograma. ¿Se trata de una distribución simétrica o asimétrica?

```python
import matplotlib.pyplot as plt

plt.hist(beers['abv'].dropna(), bins=30, edgecolor='black')
plt.xlabel('Nivel de alcohol (ABV)')
plt.ylabel('Frecuencia')
plt.title('Distribución del nivel de alcohol de las cervezas')
plt.show()
```

```python
print(beers['abv'].skew()) #este es de pnadas y está protegido frente a nulos

'''El histograma de la variable abv presenta una distribución asimétrica positiva,
 ya que el coeficiente de asimetría (skew) es 0,96. Esto indica que la distribución 
 tiene una cola hacia los valores más altos de graduación alcohólica. Por tanto, 
 predominan los valores relativamente bajos o intermedios de alcohol, mientras que
unas pocas cervezas presentan graduaciones más elevadas. La distribución no es simétrica.'''
```

## Ejercicio 23
Consigue en un DataFrame de todas las cervezas en cuyas cervecerías al menos haya una cerveza con un `abv` mínimo 0.07.

```python
# 1. Crear un DataFrame con la cervecería, la cerveza y el ABV
beers_2 = juntos[['name_y', 'name_x', 'abv']].copy()

# 2. Obtener las cervecerías que tienen alguna cerveza con ABV >= 0.07
tienda_cerve = beers_2.loc[
    beers_2['abv'] >= 0.07, 'brewery_id'
].unique()

# 3. Seleccionar TODAS las cervezas pertenecientes a esas cervecerías
juntos.loc[
    juntos['brewery_id'].isin(tienda_cerve), 'name_x'
]


```

## Ejercicio 24
Teniendo en cuenta la media de amargor de todas las cervezas, obtén una tabla con las cervezas cuyos estados tengan una media de amargor en cerveza por encima de la media

```python
media_global = juntos['ibu'].mean()
print(media_global)
media_estado = juntos.groupby('state')['ibu'].mean()
print(media_estado)
estados_superiores = media_estado[media_estado > media_global].index
print(estados_superiores)
```

```python
resultado = juntos.loc[
    juntos['state'].isin(estados_superiores)
]

print(resultado)
```
